```python
def predict_future_race(
    race_entries,
    model=final_model,
    threshold=0.65
):
    """
    Predict podium probability for an upcoming F1 race.

    race_entries must contain:
        driverId
        driver_name
        grid
        qualifying_position
        qualifying_missing
    """

    prediction_rows = []

    for _, entry in race_entries.iterrows():

        driver_id = entry["driverId"]

        # Driver's historical races before the upcoming race
        driver_history = completed_results[
            completed_results["driverId"] == driver_id
        ].sort_values("date")

        # ------------------------------------------------
        # Driver historical features
        # ------------------------------------------------

        if len(driver_history) == 0:

            previous_finish_num = 8
            previous_podium = 0
            previous_points = 0
            recent_5_podiums = 0
            recent_5_points = 0

        else:

            last_race = driver_history.iloc[-1]

            previous_finish_num = last_race["positionOrder"]
            previous_podium = last_race["podium"]
            previous_points = last_race["points"]

            recent_5 = driver_history.tail(5)

            recent_5_podiums = recent_5["podium"].sum()
            recent_5_points = recent_5["points"].sum()

        # ------------------------------------------------
        # Constructor historical features
        # ------------------------------------------------

        constructor_id = entry["constructorId"]

        constructor_history = completed_results[
            completed_results["constructorId"] == constructor_id
        ].sort_values("date")

        if len(constructor_history) == 0:

            constructor_recent_5_podiums = 0
            constructor_recent_5_points = 0

        else:

            constructor_recent_5 = constructor_history.tail(5)

            constructor_recent_5_podiums = (
                constructor_recent_5["podium"].sum()
            )

            constructor_recent_5_points = (
                constructor_recent_5["points"].sum()
            )

        # ------------------------------------------------
        # Create model input
        # ------------------------------------------------

        prediction_rows.append({
            "driverId": driver_id,
            "driver_name": entry["driver_name"],
            "grid": entry["grid"],
            "qualifying_position": entry["qualifying_position"],
            "qualifying_missing": entry["qualifying_missing"],
            "previous_finish_num": previous_finish_num,
            "previous_podium": previous_podium,
            "previous_points": previous_points,
            "recent_5_podiums": recent_5_podiums,
            "recent_5_points": recent_5_points,
            "constructor_recent_5_podiums": constructor_recent_5_podiums,
            "constructor_recent_5_points": constructor_recent_5_points
        })

    prediction_df = pd.DataFrame(prediction_rows)

    # Exact feature order used during training
    X_future = prediction_df[feature_columns]

    # Probability of podium
    probabilities = model.predict_proba(X_future)[:, 1]

    prediction_df["podium_probability"] = probabilities

    prediction_df["predicted_podium"] = (
        prediction_df["podium_probability"] >= threshold
    ).astype(int)

    # Highest probability first
    prediction_df = prediction_df.sort_values(
        "podium_probability",
        ascending=False
    ).reset_index(drop=True)

    # Predicted position among Top 3
    prediction_df["predicted_podium_position"] = (
        prediction_df.index + 1
    )
    print("Future prediction function created successfully!")

    return prediction_df
```

```python
def predict_future_race(race_entries, model, threshold=0.65):

    # --------------------------------------------------
    # 1. Get actual model object and feature list
    # --------------------------------------------------
    if isinstance(model, dict):
        model_object = model["model"]
        feature_columns = model["features"]

        # Use stored threshold if available
        threshold = model.get("threshold", threshold)
    else:
        model_object = model
        feature_columns = [
            "grid",
            "qualifying_position",
            "qualifying_missing",
            "previous_finish_num",
            "previous_podium",
            "previous_points",
            "recent_5_podiums",
            "recent_5_points",
            "constructor_recent_5_podiums",
            "constructor_recent_5_points"
        ]

    # --------------------------------------------------
    # 2. Copy data
    # --------------------------------------------------
    prediction_df = race_entries.copy()

    # --------------------------------------------------
    # 3. Check required features
    # --------------------------------------------------
    missing_features = [
        feature for feature in feature_columns
        if feature not in prediction_df.columns
    ]

    if missing_features:
        raise ValueError(
            f"Missing required features: {missing_features}"
        )

    print("✅ All required features available")

    # --------------------------------------------------
    # 4. Prepare model input
    # --------------------------------------------------
    X_future = prediction_df[feature_columns].copy()

    # Make sure values are numeric
    X_future = X_future.apply(pd.to_numeric, errors="coerce")

    # Check for missing values
    if X_future.isnull().any().any():
        print("⚠️ Missing values found:")
        print(X_future.isnull().sum())
        raise ValueError("Model input contains missing values.")

    # --------------------------------------------------
    # 5. Predict probabilities
    # --------------------------------------------------
    probabilities = model_object.predict_proba(X_future)[:, 1]

    # --------------------------------------------------
    # 6. Add prediction columns
    # --------------------------------------------------
    prediction_df["podium_probability"] = probabilities
    prediction_df["podium_probability_percent"] = probabilities * 100

    prediction_df["predicted_podium"] = (
        prediction_df["podium_probability"] >= threshold
    ).astype(int)

    prediction_df["predicted_result"] = prediction_df[
        "predicted_podium"
    ].map({
        1: "PODIUM",
        0: "NO PODIUM"
    })

    # --------------------------------------------------
    # 7. Sort by probability
    # --------------------------------------------------
    prediction_df = prediction_df.sort_values(
        "podium_probability",
        ascending=False
    ).reset_index(drop=True)

    # --------------------------------------------------
    # 8. Predicted podium position
    # --------------------------------------------------
    prediction_df["predicted_podium_position"] = None

    podium_indices = prediction_df.index[
        prediction_df["predicted_podium"] == 1
    ]

    for position, idx in enumerate(podium_indices, start=1):
        prediction_df.loc[idx, "predicted_podium_position"] = position

    # --------------------------------------------------
    # 9. Return useful columns
    # --------------------------------------------------
    output_columns = [
        "driverId",
        "driver_name",
        "constructorId",
        "grid",
        "qualifying_position",
        "podium_probability",
        "podium_probability_percent",
        "predicted_podium",
        "predicted_result",
        "predicted_podium_position"
    ]

    # Keep only columns that actually exist
    output_columns = [
        col for col in output_columns
        if col in prediction_df.columns
    ]

    return prediction_df[output_columns]
```

```python
def predict_podium(race_data, model_package):
    """
    Predict podium probabilities for a race.

    Parameters:
        race_data: DataFrame containing all required V2 features
        model_package: Saved model package containing model + features

    Returns:
        DataFrame with driver names and podium probabilities
    """

    model = model_package["model"]
    features = model_package["features"]

    X = race_data[features]

    predictions = race_data[
        ["driverId", "driver_name"]
    ].copy()

    predictions["podium_probability_percent"] = (
        model.predict_proba(X)[:, 1] * 100
    )

    predictions = predictions.sort_values(
        "podium_probability_percent",
        ascending=False
    ).reset_index(drop=True)

    predictions["predicted_podium_position"] = 0
    predictions.loc[:2, "predicted_podium_position"] = [1, 2, 3]

    predictions["podium_probability_percent"] = (
        predictions["podium_probability_percent"].round(2)
    )

    return predictions
```
